```python
from IPython.display import HTML
from base64 import b64encode
import os

# Updated to the correct output path from the last run
save_path = '/content/YOLOv8-DeepSORT-Object-Tracking/runs/detect/train9/test1.mp4'

# Compressed video path
compressed_path = "/content/result_compressed.mp4"

if os.path.exists(save_path):
    print(f"Processing {save_path}...")
    # Compress video for browser playback
    os.system(f"ffmpeg -y -i {save_path} -vcodec libx264 {compressed_path} > /dev/null 2>&1")

    if os.path.exists(compressed_path):
        mp4 = open(compressed_path,'rb').read()
        data_url = "data:video/mp4;base64," + b64encode(mp4).decode()
        display(HTML(f"""
        <video width=600 controls>
              <source src='{data_url}' type='video/mp4'>
        </video>
        """))
    else:
        print(f"Error: Compression failed.")
else:
    print(f"Error: Input video not found at {save_path}.")
```
